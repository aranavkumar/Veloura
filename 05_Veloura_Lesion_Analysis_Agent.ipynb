```python

def segment_lesion(image_bgr):
    """
    Experimental lesion segmentation using OpenCV GrabCut.
    Returns a binary mask and basic segmentation measurements.
    """

    if image_bgr is None:
        return None, {
            "status": "LOW_RELIABILITY",
            "reason": "IMAGE_NOT_AVAILABLE"
        }

    height, width = image_bgr.shape[:2]

    # Initialize GrabCut mask
    mask = np.zeros((height, width), np.uint8)

    # Central rectangle: initial foreground search region
    x = int(width * 0.15)
    y = int(height * 0.15)
    rect_width = int(width * 0.70)
    rect_height = int(height * 0.70)

    rect = (x, y, rect_width, rect_height)

    background_model = np.zeros((1, 65), np.float64)
    foreground_model = np.zeros((1, 65), np.float64)

    try:
        cv2.grabCut(
            image_bgr,
            mask,
            rect,
            background_model,
            foreground_model,
            5,
            cv2.GC_INIT_WITH_RECT
        )

        # Definite and probable foreground pixels
        binary_mask = np.where(
            (mask == cv2.GC_FGD) |
            (mask == cv2.GC_PR_FGD),
            255,
            0
        ).astype(np.uint8)

        # Keep the largest connected foreground component
        contours, _ = cv2.findContours(
            binary_mask,
            cv2.RETR_EXTERNAL,
            cv2.CHAIN_APPROX_SIMPLE
        )

        if not contours:
            return binary_mask, {
                "status": "LOW_RELIABILITY",
                "reason": "NO_FOREGROUND_COMPONENT"
            }

        largest_contour = max(contours, key=cv2.contourArea)

        cleaned_mask = np.zeros_like(binary_mask)

        cv2.drawContours(
            cleaned_mask,
            [largest_contour],
            -1,
            255,
            thickness=cv2.FILLED
        )

        area = cv2.contourArea(largest_contour)
        image_area = height * width
        area_fraction = area / image_area

        # Basic engineering reliability checks
        reasons = []

        if area_fraction < 0.01:
            reasons.append("SEGMENT_TOO_SMALL")

        if area_fraction > 0.65:
            reasons.append("SEGMENT_TOO_LARGE")

        if area < 100:
            reasons.append("INSUFFICIENT_SEGMENT_AREA")

        if len(largest_contour) < 5:
            reasons.append("INSUFFICIENT_BOUNDARY_POINTS")

        status = (
            "LOW_RELIABILITY"
            if reasons
            else "REVIEW_REQUIRED"
        )

        return cleaned_mask, {
            "status": status,
            "area_pixels": int(area),
            "area_fraction": round(area_fraction, 4),
            "reason": reasons
        }

    except cv2.error as error:
        return None, {
            "status": "LOW_RELIABILITY",
            "reason": "GRABCUT_FAILED",
            "error": str(error)
        }


print("GrabCut segmentation function created successfully!")
```

```python

def segment_lesion_v2(image_bgr):
    """
    Experimental seeded GrabCut segmentation.
    The output is a candidate mask, not a validated lesion boundary.
    """

    if image_bgr is None:
        return None, {
            "status": "LOW_RELIABILITY",
            "reason": "IMAGE_NOT_AVAILABLE"
        }

    height, width = image_bgr.shape[:2]

    # Initialize every pixel as definite background
    mask = np.full(
        (height, width),
        cv2.GC_BGD,
        dtype=np.uint8
    )

    # Define an inner rectangle as probable foreground
    margin_x = int(width * 0.06)
    margin_y = int(height * 0.06)

    mask[
        margin_y:height-margin_y,
        margin_x:width-margin_x
    ] = cv2.GC_PR_FGD

    # Seed a central ellipse as definite foreground
    center = (width // 2, height // 2)

    axes = (
        max(1, int(width * 0.14)),
        max(1, int(height * 0.14))
    )

    cv2.ellipse(
        mask,
        center,
        axes,
        0,
        0,
        360,
        cv2.GC_FGD,
        thickness=-1
    )

    background_model = np.zeros((1, 65), np.float64)
    foreground_model = np.zeros((1, 65), np.float64)

    try:
        cv2.grabCut(
            image_bgr,
            mask,
            None,
            background_model,
            foreground_model,
            5,
            cv2.GC_INIT_WITH_MASK
        )

        binary_mask = np.where(
            (mask == cv2.GC_FGD) |
            (mask == cv2.GC_PR_FGD),
            255,
            0
        ).astype(np.uint8)

        contours, _ = cv2.findContours(
            binary_mask,
            cv2.RETR_EXTERNAL,
            cv2.CHAIN_APPROX_SIMPLE
        )

        if not contours:
            return binary_mask, {
                "status": "LOW_RELIABILITY",
                "reason": "NO_FOREGROUND_COMPONENT"
            }

        largest_contour = max(
            contours,
            key=cv2.contourArea
        )

        cleaned_mask = np.zeros_like(binary_mask)

        cv2.drawContours(
            cleaned_mask,
            [largest_contour],
            -1,
            255,
            thickness=cv2.FILLED
        )

        area = cv2.contourArea(largest_contour)
        area_fraction = area / (height * width)

        reasons = []

        if area_fraction < 0.01:
            reasons.append("SEGMENT_TOO_SMALL")

        if area_fraction > 0.65:
            reasons.append("SEGMENT_TOO_LARGE")

        if area < 100:
            reasons.append("INSUFFICIENT_SEGMENT_AREA")

        if len(largest_contour) < 5:
            reasons.append("INSUFFICIENT_BOUNDARY_POINTS")

        return cleaned_mask, {
            "status": (
                "LOW_RELIABILITY"
                if reasons
                else "CANDIDATE_REVIEW"
            ),
            "area_pixels": int(area),
            "area_fraction": round(area_fraction, 4),
            "reason": reasons
        }

    except cv2.error as error:
        return None, {
            "status": "LOW_RELIABILITY",
            "reason": "GRABCUT_FAILED",
            "error": str(error)
        }


print("Seeded GrabCut v2 is ready!")
```

```python

def evaluate_segmentation_reliability(mask):
    """
    Engineering quality checks for a candidate segmentation mask.
    These checks do not establish clinical segmentation accuracy.
    """

    if mask is None:
        return {
            "status": "LOW_RELIABILITY",
            "reliability_flags": ["MASK_NOT_AVAILABLE"]
        }

    height, width = mask.shape
    image_area = height * width

    binary = (mask > 0).astype(np.uint8)

    area = int(binary.sum())
    area_fraction = area / image_area

    flags = []

    if area == 0:
        flags.append("EMPTY_MASK")

    if area_fraction < 0.01:
        flags.append("VERY_SMALL_MASK")

    if area_fraction > 0.40:
        flags.append("LARGE_FOREGROUND_REGION")

    # Check whether the mask touches the image boundary
    border_pixels = np.concatenate([
        binary[0, :],
        binary[-1, :],
        binary[:, 0],
        binary[:, -1]
    ])

    if np.any(border_pixels > 0):
        flags.append("MASK_TOUCHES_IMAGE_BORDER")

    # Count connected components
    component_count, labels, stats, centroids = (
        cv2.connectedComponentsWithStats(binary, 8)
    )

    foreground_components = component_count - 1

    if foreground_components > 1:
        flags.append("MULTIPLE_FOREGROUND_COMPONENTS")

    # This is only a technical review flag.
    # It cannot determine whether the lesion boundary is correct.
    return {
        "status": "REVIEW_REQUIRED",
        "area_pixels": area,
        "area_fraction": round(area_fraction, 4),
        "foreground_components": foreground_components,
        "reliability_flags": flags,
        "automatic_acceptance": False
    }


print("Segmentation reliability evaluator created!")
```

```python

def extract_lesion_features(image_bgr, mask):
    """
    Extract experimental visual features from a candidate lesion mask.

    All measurements are provisional. They are not clinical ABCD scores.
    """

    if image_bgr is None or mask is None:
        return {
            "feature_status": "UNAVAILABLE",
            "reason": "IMAGE_OR_MASK_MISSING"
        }

    reliability = evaluate_segmentation_reliability(mask)

    if reliability["area_pixels"] < 100:
        return {
            "feature_status": "UNAVAILABLE",
            "reason": "INSUFFICIENT_MASK_AREA",
            "segmentation_reliability": reliability
        }

    # Ensure binary mask
    binary = (mask > 0).astype(np.uint8)

    # 1. Geometric asymmetry estimate
    moments = cv2.moments(binary)

    if moments["m00"] == 0:
        asymmetry = None
    else:
        center_x = int(moments["m10"] / moments["m00"])
        center_y = int(moments["m01"] / moments["m00"])

        x1 = max(0, 2 * center_x - binary.shape[1])
        x2 = min(binary.shape[1], 2 * center_x + 1)

        y1 = max(0, 2 * center_y - binary.shape[0])
        y2 = min(binary.shape[0], 2 * center_y + 1)

        cropped = binary[y1:y2, x1:x2]

        # Compare the mask with its horizontal and vertical reflections.
        horizontal_difference = np.mean(
            np.abs(cropped - np.fliplr(cropped))
        )

        vertical_difference = np.mean(
            np.abs(cropped - np.flipud(cropped))
        )

        asymmetry = float(
            (horizontal_difference + vertical_difference) / 2
        )

    # 2. Boundary irregularity estimate
    contours, _ = cv2.findContours(
        binary,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    if contours:
        contour = max(contours, key=cv2.contourArea)

        perimeter = cv2.arcLength(contour, True)
        area = cv2.contourArea(contour)

        if perimeter > 0 and area > 0:
            circularity = (
                4 * np.pi * area / (perimeter ** 2)
            )
            circularity = float(
                np.clip(circularity, 0, 1)
            )
            boundary_irregularity = 1 - circularity
        else:
            circularity = None
            boundary_irregularity = None
    else:
        circularity = None
        boundary_irregularity = None

    # 3. Colour variation within the candidate mask
    hsv = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2HSV)

    selected_pixels = hsv[binary > 0]

    if len(selected_pixels) > 0:
        hue_std = float(np.std(selected_pixels[:, 0]))
        saturation_std = float(np.std(selected_pixels[:, 1]))
        value_std = float(np.std(selected_pixels[:, 2]))
    else:
        hue_std = None
        saturation_std = None
        value_std = None

    # Never represent these measurements as trusted when
    # segmentation has not been independently validated.
    return {
        "feature_status": "EXPERIMENTAL_UNVALIDATED",
        "asymmetry_estimate": (
            round(asymmetry, 4)
            if asymmetry is not None else None
        ),
        "boundary_irregularity_estimate": (
            round(boundary_irregularity, 4)
            if boundary_irregularity is not None else None
        ),
        "circularity_estimate": (
            round(circularity, 4)
            if circularity is not None else None
        ),
        "colour_variation": {
            "hue_std": round(hue_std, 2) if hue_std is not None else None,
            "saturation_std": round(saturation_std, 2) if saturation_std is not None else None,
            "brightness_std": round(value_std, 2) if value_std is not None else None
        },
        "segmentation_reliability": reliability
    }


print("Lesion feature extraction function created!")
```

```python

def extract_lesion_features(image_bgr, mask):
    """
    Extract experimental geometric and colour features.
    Measurements are not clinically validated.
    """

    if image_bgr is None or mask is None:
        return {
            "feature_status": "UNAVAILABLE",
            "reason": "IMAGE_OR_MASK_MISSING"
        }

    binary = (mask > 0).astype(np.uint8)
    reliability = evaluate_segmentation_reliability(mask)

    if binary.sum() < 100:
        return {
            "feature_status": "UNAVAILABLE",
            "reason": "INSUFFICIENT_MASK_AREA",
            "segmentation_reliability": reliability
        }

    height, width = binary.shape
    moments = cv2.moments(binary)

    if moments["m00"] == 0:
        asymmetry = None
    else:
        cx = int(moments["m10"] / moments["m00"])
        cy = int(moments["m01"] / moments["m00"])

        # Use an equally sized region around the centroid
        radius_x = min(cx, width - 1 - cx)
        radius_y = min(cy, height - 1 - cy)
        radius = min(radius_x, radius_y)

        if radius < 2:
            asymmetry = None
        else:
            crop = binary[
                cy-radius:cy+radius+1,
                cx-radius:cx+radius+1
            ].astype(np.float32)

            horizontal_difference = np.mean(
                np.abs(crop - np.fliplr(crop))
            )

            vertical_difference = np.mean(
                np.abs(crop - np.flipud(crop))
            )

            asymmetry = float(
                (horizontal_difference + vertical_difference) / 2
            )

    # Boundary irregularity estimate
    contours, _ = cv2.findContours(
        binary,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    if contours:
        contour = max(contours, key=cv2.contourArea)
        perimeter = cv2.arcLength(contour, True)
        area = cv2.contourArea(contour)

        if perimeter > 0 and area > 0:
            circularity = float(np.clip(
                4 * np.pi * area / (perimeter ** 2),
                0, 1
            ))
            boundary_irregularity = 1 - circularity
        else:
            circularity = None
            boundary_irregularity = None
    else:
        circularity = None
        boundary_irregularity = None

    # Colour variation inside candidate mask
    hsv = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2HSV)
    selected = hsv[binary > 0]

    if len(selected) > 0:
        # Hue is circular, so use circular dispersion rather
        # than ordinary standard deviation.
        hue_angles = selected[:, 0].astype(float) * (
            2 * np.pi / 180
        )

        mean_sin = np.mean(np.sin(hue_angles))
        mean_cos = np.mean(np.cos(hue_angles))
        resultant_length = np.sqrt(
            mean_sin**2 + mean_cos**2
        )

        hue_circular_dispersion = float(
            1 - np.clip(resultant_length, 0, 1)
        )

        saturation_std = float(np.std(selected[:, 1]))
        brightness_std = float(np.std(selected[:, 2]))
    else:
        hue_circular_dispersion = None
        saturation_std = None
        brightness_std = None

    return {
        "feature_status": "EXPERIMENTAL_UNVALIDATED",
        "asymmetry_estimate": (
            round(asymmetry, 4)
            if asymmetry is not None else None
        ),
        "boundary_irregularity_estimate": (
            round(boundary_irregularity, 4)
            if boundary_irregularity is not None else None
        ),
        "circularity_estimate": (
            round(circularity, 4)
            if circularity is not None else None
        ),
        "colour_variation": {
            "hue_circular_dispersion": (
                round(hue_circular_dispersion, 4)
                if hue_circular_dispersion is not None else None
            ),
            "saturation_std": (
                round(saturation_std, 2)
                if saturation_std is not None else None
            ),
            "brightness_std": (
                round(brightness_std, 2)
                if brightness_std is not None else None
            )
        },
        "segmentation_reliability": reliability
    }


print("Corrected feature extraction function loaded!")
```

```python

class LesionAnalysisAgent:

    def __init__(self):
        self.agent_name = "Veloura Lesion Analysis Agent"

    def analyze(self, image_path):

        image_id = os.path.splitext(
            os.path.basename(image_path)
        )[0]

        image_bgr = cv2.imread(image_path)

        if image_bgr is None:
            return {
                "agent": self.agent_name,
                "image_id": image_id,
                "status": "ERROR",
                "error": "IMAGE_COULD_NOT_BE_OPENED",
                "clinical_interpretation": None
            }

        # Step 1: Candidate segmentation
        mask, segmentation_result = segment_lesion(image_bgr)

        if mask is None:
            return {
                "agent": self.agent_name,
                "image_id": image_id,
                "status": "LOW_RELIABILITY",
                "segmentation": segmentation_result,
                "features": None,
                "clinical_interpretation": None
            }

        # Step 2: Extract experimental features
        features = extract_lesion_features(
            image_bgr,
            mask
        )

        # Step 3: Build transparent warnings
        warnings = [
            "SEGMENTATION_NOT_VALIDATED_AGAINST_EXPERT_MASKS",
            "VISUAL_FEATURES_ARE_EXPERIMENTAL",
            "NO_CLINICAL_RISK_SCORE_GENERATED",
            "MANUAL_REVIEW_REQUIRED"
        ]

        flags = features.get(
            "segmentation_reliability", {}
        ).get("reliability_flags", [])

        warnings.extend(flags)

        return {
            "agent": self.agent_name,
            "image_id": image_id,
            "status": "REVIEW_REQUIRED",
            "segmentation_method": "GRABCUT_V1",
            "segmentation": segmentation_result,
            "features": features,
            "evidence_summary": {
                "asymmetry_estimate": features.get(
                    "asymmetry_estimate"
                ),
                "boundary_irregularity_estimate": features.get(
                    "boundary_irregularity_estimate"
                ),
                "circularity_estimate": features.get(
                    "circularity_estimate"
                ),
                "colour_variation": features.get(
                    "colour_variation"
                )
            },
            "reliability_flags": flags,
            "warnings": warnings,
            "clinical_interpretation": None,
            "automatic_acceptance": False
        }


lesion_agent = LesionAnalysisAgent()

print("Lesion Analysis Agent initialized!")
```
