```python

# ============================================================
# VELOURA PHASE 11
# CELL 2: AGENT SCHEMA DISCOVERY & DATA CONTRACT AUDIT
# ============================================================

import json
import pandas as pd
from pathlib import Path

PROJECT_DIR = Path(PROJECT_DIR)

PHASE11_OUTPUT = (
    PROJECT_DIR / "outputs/workflow_orchestration"
)

PHASE11_REPORTS = (
    PROJECT_DIR / "reports/workflow_orchestration"
)

target_ids = set(REFERENCE_IDS)

# ------------------------------------------------------------
# 1. HELPER: RECURSIVE RECORD DISCOVERY
# ------------------------------------------------------------

def discover_image_records(data, target_ids, results=None):
    """
    Recursively discover dictionaries containing a target image_id.
    Does not modify source data.
    """

    if results is None:
        results = {}

    if isinstance(data, dict):

        image_id = data.get("image_id")

        if image_id in target_ids:
            results.setdefault(image_id, data)

        for value in data.values():
            if isinstance(value, (dict, list)):
                discover_image_records(
                    value, target_ids, results
                )

    elif isinstance(data, list):

        for item in data:
            if isinstance(item, (dict, list)):
                discover_image_records(
                    item, target_ids, results
                )

    return results


# ------------------------------------------------------------
# 2. HELPER: STRUCTURE SUMMARY
# ------------------------------------------------------------

def describe_structure(data, depth=0, max_depth=2):

    if depth > max_depth:
        return "MAX_DEPTH_REACHED"

    if isinstance(data, dict):

        summary = {}

        for key, value in list(data.items())[:30]:

            if isinstance(value, dict):
                summary[key] = describe_structure(
                    value, depth + 1, max_depth
                )

            elif isinstance(value, list):
                summary[key] = {
                    "type": "list",
                    "length": len(value),
                    "first_item_type": (
                        type(value[0]).__name__
                        if value else None
                    )
                }

            else:
                summary[key] = type(value).__name__

        return summary

    elif isinstance(data, list):

        return {
            "type": "list",
            "length": len(data),
            "first_item_type": (
                type(data[0]).__name__
                if data else None
            )
        }

    return type(data).__name__


# ------------------------------------------------------------
# 3. INSPECT ALL UPSTREAM DATASETS
# ------------------------------------------------------------

print("=" * 75)
print("AGENT DATA STRUCTURE DISCOVERY")
print("=" * 75)

schema_results = []
MATCHED_RECORDS = {}

for agent_name, data in UPSTREAM_DATA.items():

    print("\n" + "-" * 75)
    print("AGENT:", agent_name)
    print("-" * 75)

    print("Top-level type:", type(data).__name__)

    if isinstance(data, dict):
        print("Top-level keys:", list(data.keys())[:40])

    elif isinstance(data, list):
        print("Top-level record count:", len(data))

    # Discover matching case records
    matches = discover_image_records(
        data,
        target_ids
    )

    MATCHED_RECORDS[agent_name] = matches

    print("Target image records found:", len(matches))

    for image_id in sorted(matches):

        record = matches[image_id]

        print("\nImage:", image_id)
        print("Top-level record fields:", list(record.keys()))

        # Display nested structure without dumping full records
        print(
            "Record structure:",
            json.dumps(
                describe_structure(record),
                indent=2
            )[:2500]
        )

        # Collect schema information
        schema_results.append({
            "agent": agent_name,
            "image_id": image_id,
            "record_fields": ", ".join(record.keys()),
            "field_count": len(record),
            "record_type": type(record).__name__
        })

    if not matches:
        print(
            "No direct image_id records discovered. "
            "This dataset may contain aggregate information."
        )


# ------------------------------------------------------------
# 4. CASE COVERAGE MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CASE COVERAGE MATRIX")
print("=" * 75)

coverage_rows = []

for agent_name in UPSTREAM_FILES:

    matches = MATCHED_RECORDS.get(agent_name, {})

    row = {"agent": agent_name}

    for image_id in REFERENCE_IDS:
        row[image_id] = (
            "FOUND" if image_id in matches else "NOT_FOUND"
        )

    row["matched_cases"] = len(matches)

    coverage_rows.append(row)

coverage_df = pd.DataFrame(coverage_rows)

print(coverage_df.to_string(index=False))

coverage_path = (
    PHASE11_REPORTS / "agent_case_coverage_matrix.csv"
)

coverage_df.to_csv(
    coverage_path,
    index=False
)


# ------------------------------------------------------------
# 5. SAFETY AND COORDINATOR DECISION INSPECTION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("SAFETY AND COORDINATOR FIELD DISCOVERY")
print("=" * 75)

decision_agents = [
    "safety_agent",
    "coordinator_integration",
    "master_coordinator"
]

for agent_name in decision_agents:

    matches = MATCHED_RECORDS.get(agent_name, {})

    print("\nAGENT:", agent_name)

    for image_id, record in matches.items():

        print("\nIMAGE:", image_id)

        # Print selected decision-related fields
        for key, value in record.items():

            key_lower = key.lower()

            if any(term in key_lower for term in [
                "status",
                "decision",
                "action",
                "safety",
                "review",
                "accept",
                "diagnos",
                "workflow",
                "gate",
                "escalat",
                "warning",
                "flag"
            ]):

                print(
                    f"{key}:",
                    json.dumps(value, default=str)[:1200]
                )


# ------------------------------------------------------------
# 6. COMMON FIELD DISCOVERY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("COMMON FIELD DISCOVERY")
print("=" * 75)

field_sets = {}

for agent_name, matches in MATCHED_RECORDS.items():

    if matches:

        first_record = next(iter(matches.values()))

        field_sets[agent_name] = set(first_record.keys())

        print(
            agent_name,
            "=>",
            sorted(first_record.keys())
        )

if field_sets:

    common_fields = set.intersection(
        *field_sets.values()
    ) if len(field_sets) > 1 else set()

    print("\nFields common to all matched agent datasets:")
    print(sorted(common_fields))


# ------------------------------------------------------------
# 7. SAVE SCHEMA DISCOVERY REPORT
# ------------------------------------------------------------

schema_df = pd.DataFrame(schema_results)

schema_path = (
    PHASE11_REPORTS / "agent_schema_discovery.csv"
)

schema_df.to_csv(
    schema_path,
    index=False
)

# Save a compact machine-readable contract discovery report
contract_report = {
    "phase": 11,
    "stage": "AGENT_SCHEMA_DISCOVERY",
    "target_image_ids": sorted(target_ids),
    "datasets_inspected": len(UPSTREAM_DATA),
    "agents": {},
    "common_top_level_fields": sorted(common_fields)
        if field_sets else []
}

for agent_name, matches in MATCHED_RECORDS.items():

    contract_report["agents"][agent_name] = {
        "matched_case_count": len(matches),
        "matched_image_ids": sorted(matches.keys()),
        "available": agent_name in UPSTREAM_DATA
    }

contract_path = (
    PHASE11_OUTPUT / "agent_schema_discovery.json"
)

with open(contract_path, "w", encoding="utf-8") as f:
    json.dump(
        contract_report,
        f,
        indent=2
    )


# ------------------------------------------------------------
# 8. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PHASE 11 — CELL 2 SUMMARY")
print("=" * 75)

print("Datasets inspected:", len(UPSTREAM_DATA))
print("Case-level coverage matrix:", coverage_path)
print("Schema discovery CSV:", schema_path)
print("Schema discovery JSON:", contract_path)

for agent_name in UPSTREAM_FILES:

    print(
        agent_name,
        ":",
        len(MATCHED_RECORDS.get(agent_name, {})),
        "/ 6 cases"
    )

print("\nSUCCESS: AGENT SCHEMA DISCOVERY COMPLETED!")
```

```python
# ============================================================================
# PHASE 11 — CELL 3
# SHARED DATA CONTRACTS & CASE BUNDLE ASSEMBLY
# ============================================================================

import json
import pandas as pd
from pathlib import Path
from datetime import datetime, timezone

print("=" * 75)
print("PHASE 11 — SHARED DATA CONTRACTS & CASE BUNDLE ASSEMBLY")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. Confirm prerequisites from Cells 1 and 2
# ---------------------------------------------------------------------------

required_globals = ["PROJECT_DIR", "OUTPUT_DIR", "REPORT_DIR",
                    "REFERENCE_IDS", "UPSTREAM_DATA", "MATCHED_RECORDS"]

missing_globals = [name for name in required_globals if name not in globals()]

if missing_globals:
    raise RuntimeError(
        f"Required variables missing. Run Cells 1 and 2 first: {missing_globals}"
    )

CASE_DATASETS = [
    "image_quality",
    "lesion_analysis",
    "reasoning_context",
    "safety_agent",
    "coordinator_integration",
    "master_coordinator",
    "evaluation_performance",
    "explainability_evidence",
    "clinician_summaries",
    "visual_evidence_index"
]

# Dataset-level information is deliberately kept separate from case records.
AGGREGATE_DATASET = UPSTREAM_DATA.get("evaluation_aggregate", {})

contract_results = []
CASE_BUNDLES = {}

def add_check(image_id, dataset, check_name, passed, detail):
    contract_results.append({
        "image_id": image_id,
        "dataset": dataset,
        "check": check_name,
        "status": "PASS" if passed else "FAIL",
        "detail": str(detail)
    })

def is_dict(value):
    return isinstance(value, dict)

def is_list(value):
    return isinstance(value, list)

def has_fields(record, fields):
    return is_dict(record) and all(field in record for field in fields)

def nested(record, *keys):
    current = record
    for key in keys:
        if not isinstance(current, dict):
            return None
        current = current.get(key)
    return current

# ---------------------------------------------------------------------------
# 2. Required source fields discovered in Cell 2
# ---------------------------------------------------------------------------

REQUIRED_FIELDS = {
    "image_quality": [
        "image_id", "status", "quality_score", "width", "height",
        "blur_score", "brightness", "contrast", "glare_fraction",
        "issues", "recommendation"
    ],
    "lesion_analysis": [
        "image_id", "status", "segmentation_method", "segmentation",
        "features", "evidence_summary", "reliability_flags",
        "warnings", "clinical_interpretation", "automatic_acceptance"
    ],
    "reasoning_context": [
        "image_id", "patient_context", "lesion_context",
        "clinical_constraints", "missing_information",
        "context_completeness"
    ],
    "safety_agent": [
        "image_id", "safety_decision", "safety_flags",
        "warnings", "clinical_safeguards"
    ],
    "coordinator_integration": [
        "image_id", "workflow_status", "safety_decision",
        "safety_gate", "workflow_permissions", "coordinator_action"
    ],
    "master_coordinator": [
        "image_id", "orchestration_status", "safety_summary",
        "workflow_permissions", "coordinator_action",
        "clinical_authority", "agent_availability"
    ],
    "evaluation_performance": [
        "image_id", "technical_evaluation", "clinical_metrics",
        "clinical_validation_status", "safety_evaluation"
    ],
    "explainability_evidence": [
        "image_id", "analysis_evidence", "safety_explanation",
        "coordinator_explanation", "evaluation_evidence",
        "evidence_provenance", "clinical_limitations"
    ],
    "clinician_summaries": [
        "image_id", "analysis_overview", "safety_review",
        "coordinator_decision", "evaluation_status",
        "clinician_review_note", "final_summary"
    ],
    "visual_evidence_index": [
        "image_id", "source_image", "original_experimental_mask",
        "cleaned_experimental_mask", "original_overlay",
        "cleaned_boundary_overlay", "visual_review_status",
        "expert_reference_mask_available",
        "expert_annotation_verified", "clinical_validation",
        "automatic_acceptance", "mandatory_human_review"
    ]
}

EXPECTED_SAFETY_FLAGS = {
    "MISSING_PATIENT_CONTEXT",
    "EXPERIMENTAL_FEATURES",
    "UNVALIDATED_SEGMENTATION",
    "MANDATORY_HUMAN_REVIEW"
}

# ---------------------------------------------------------------------------
# 3. Validate each case and construct its source-preserving bundle
# ---------------------------------------------------------------------------

for image_id in REFERENCE_IDS:

    case_records = {}

    # Retrieve all source records without modifying them.
    for dataset in CASE_DATASETS:
        record = MATCHED_RECORDS.get(dataset, {}).get(image_id)
        case_records[dataset] = record

        add_check(
            image_id, dataset, "record_present",
            is_dict(record),
            "Case-level source record found" if is_dict(record)
            else "Missing or invalid case-level source record"
        )

        if not is_dict(record):
            continue

        add_check(
            image_id, dataset, "image_id_matches",
            record.get("image_id") == image_id,
            f"Source image_id={record.get('image_id')}"
        )

        missing_fields = [
            field for field in REQUIRED_FIELDS[dataset]
            if field not in record
        ]

        add_check(
            image_id, dataset, "required_fields_present",
            len(missing_fields) == 0,
            "All required fields present" if not missing_fields
            else f"Missing fields: {missing_fields}"
        )

    # -----------------------------------------------------------------------
    # 4. Safety Agent invariants
    # -----------------------------------------------------------------------

    safety = case_records.get("safety_agent") or {}
    safety_flags = safety.get("safety_flags", [])
    safeguards = safety.get("clinical_safeguards", {})

    add_check(
        image_id, "safety_agent", "expected_safety_flags",
        is_list(safety_flags) and set(safety_flags) == EXPECTED_SAFETY_FLAGS,
        f"Observed flags: {safety_flags}"
    )

    add_check(
        image_id, "safety_agent", "human_review_required",
        safeguards.get("human_review_required") is True,
        f"Value: {safeguards.get('human_review_required')}"
    )

    add_check(
        image_id, "safety_agent", "automatic_acceptance_disabled",
        safeguards.get("automatic_acceptance") is False,
        f"Value: {safeguards.get('automatic_acceptance')}"
    )

    add_check(
        image_id, "safety_agent", "independent_diagnosis_disabled",
        safeguards.get("independent_diagnosis") is False,
        f"Value: {safeguards.get('independent_diagnosis')}"
    )

    add_check(
        image_id, "safety_agent", "experimental_evidence_unvalidated",
        safeguards.get("experimental_evidence_validated") is False,
        f"Value: {safeguards.get('experimental_evidence_validated')}"
    )

    # -----------------------------------------------------------------------
    # 5. Coordinator gate and permissions
    # -----------------------------------------------------------------------

    coordinator = case_records.get("coordinator_integration") or {}
    gate = coordinator.get("safety_gate", {})
    permissions = coordinator.get("workflow_permissions", {})

    add_check(
        image_id, "coordinator_integration", "safety_gate_closed",
        gate.get("passed") is False,
        f"Gate passed: {gate.get('passed')}"
    )

    add_check(
        image_id, "coordinator_integration", "workflow_blocked",
        coordinator.get("workflow_status") == "BLOCKED_PENDING_HUMAN_REVIEW",
        f"Workflow status: {coordinator.get('workflow_status')}"
    )

    add_check(
        image_id, "coordinator_integration", "clinician_escalation",
        coordinator.get("coordinator_action") == "ESCALATE_TO_CLINICIAN",
        f"Action: {coordinator.get('coordinator_action')}"
    )

    add_check(
        image_id, "coordinator_integration", "clinical_permissions_restricted",
        (
            permissions.get("allow_automated_clinical_clearance") is False
            and permissions.get("allow_independent_diagnosis") is False
            and permissions.get("allow_automatic_acceptance") is False
            and permissions.get("continue_to_human_review") is True
        ),
        f"Permissions: {permissions}"
    )

    add_check(
        image_id, "coordinator_integration", "safety_flags_preserved",
        set(gate.get("flags", [])) == EXPECTED_SAFETY_FLAGS,
        f"Coordinator flags: {gate.get('flags')}"
    )

    # -----------------------------------------------------------------------
    # 6. Master Coordinator invariants
    # -----------------------------------------------------------------------

    master = case_records.get("master_coordinator") or {}
    master_safety = master.get("safety_summary", {})
    master_permissions = master.get("workflow_permissions", {})
    clinical_authority = master.get("clinical_authority", {})

    add_check(
        image_id, "master_coordinator", "orchestration_blocked",
        master.get("orchestration_status") == "BLOCKED_PENDING_HUMAN_REVIEW",
        f"Status: {master.get('orchestration_status')}"
    )

    add_check(
        image_id, "master_coordinator", "master_safety_gate_closed",
        master_safety.get("safety_gate_passed") is False,
        f"Gate passed: {master_safety.get('safety_gate_passed')}"
    )

    add_check(
        image_id, "master_coordinator", "master_human_review_required",
        master_safety.get("human_review_required") is True
        and master_permissions.get("human_review_required") is True,
        "Human review requirement checked in both sections"
    )

    add_check(
        image_id, "master_coordinator", "master_permissions_restricted",
        (
            master_permissions.get("automatic_acceptance") is False
            and master_permissions.get("independent_diagnosis") is False
            and master_permissions.get("automated_clinical_clearance") is False
        ),
        f"Permissions: {master_permissions}"
    )

    add_check(
        image_id, "master_coordinator", "clinical_authority_preserved",
        (
            clinical_authority.get("autonomous_diagnosis") is False
            and clinical_authority.get("autonomous_treatment_recommendation") is False
            and clinical_authority.get("final_decision_owner") == "QUALIFIED_CLINICIAN"
        ),
        f"Clinical authority: {clinical_authority}"
    )

    # -----------------------------------------------------------------------
    # 7. Cross-agent consistency checks
    # -----------------------------------------------------------------------

    add_check(
        image_id, "cross_agent", "safety_decision_consistent",
        (
            safety.get("safety_decision")
            == coordinator.get("safety_decision")
            == master_safety.get("safety_decision")
        ),
        "Safety decisions compared across three source agents"
    )

    add_check(
        image_id, "cross_agent", "coordinator_action_consistent",
        coordinator.get("coordinator_action") == master.get("coordinator_action"),
        "Coordinator action compared between Phase 7 and master coordinator"
    )

    add_check(
        image_id, "cross_agent", "safety_flags_consistent",
        (
            set(safety.get("safety_flags", []))
            == set(gate.get("flags", []))
            == set(master_safety.get("safety_flags", []))
        ),
        "Safety flags compared across all three safety/workflow records"
    )

    # -----------------------------------------------------------------------
    # 8. Evaluation and visual evidence limitations
    # -----------------------------------------------------------------------

    evaluation = case_records.get("evaluation_performance") or {}
    metrics = evaluation.get("clinical_metrics", {})

    unavailable_metrics = [
        name for name, metric in metrics.items()
        if isinstance(metric, dict) and metric.get("value") is None
    ]

    add_check(
        image_id, "evaluation_performance", "clinical_metrics_not_invented",
        len(unavailable_metrics) == len(metrics) and len(metrics) > 0,
        f"Metrics with unavailable values: {unavailable_metrics}"
    )

    add_check(
        image_id, "evaluation_performance", "clinical_validation_not_claimed",
        evaluation.get("clinical_validation_status") == "NOT_CLINICALLY_VALIDATED",
        f"Validation status: {evaluation.get('clinical_validation_status')}"
    )

    visual = case_records.get("visual_evidence_index") or {}

    add_check(
        image_id, "visual_evidence_index", "expert_mask_not_claimed",
        (
            visual.get("expert_reference_mask_available") is False
            and visual.get("expert_annotation_verified") is False
        ),
        "Expert reference mask and annotation verification remain unavailable"
    )

    add_check(
        image_id, "visual_evidence_index", "visual_human_review_preserved",
        (
            visual.get("mandatory_human_review") is True
            and visual.get("automatic_acceptance") is False
        ),
        "Visual evidence remains subject to mandatory human review"
    )

    # -----------------------------------------------------------------------
    # 9. Assemble bundle with original source records and provenance
    # -----------------------------------------------------------------------

    CASE_BUNDLES[image_id] = {
        "image_id": image_id,
        "bundle_schema_version": "1.0",
        "created_at_utc": datetime.now(timezone.utc).isoformat(),
        "bundle_status": "CONTRACT_AUDIT_PENDING",
        "source_datasets": case_records,
        "dataset_level_context": {
            "evaluation_aggregate": AGGREGATE_DATASET
        },
        "safety_policy": {
            "automatic_acceptance_enabled": False,
            "independent_diagnosis_enabled": False,
            "automated_clinical_clearance_enabled": False,
            "mandatory_human_review": True,
            "final_decision_owner": "QUALIFIED_CLINICIAN"
        }
    }

# ---------------------------------------------------------------------------
# 10. Finalize contract audit and bundle statuses
# ---------------------------------------------------------------------------

contract_df = pd.DataFrame(contract_results)

failed_checks = contract_df[contract_df["status"] == "FAIL"].copy()
passed_checks = contract_df[contract_df["status"] == "PASS"].copy()

for image_id, bundle in CASE_BUNDLES.items():
    case_checks = contract_df[contract_df["image_id"] == image_id]
    case_failed = (case_checks["status"] == "FAIL").any()

    bundle["bundle_status"] = (
        "CONTRACT_VALIDATED" if not case_failed
        else "CONTRACT_REVIEW_REQUIRED"
    )

# ---------------------------------------------------------------------------
# 11. Save outputs
# ---------------------------------------------------------------------------

OUTPUT_DIR = Path(OUTPUT_DIR)
REPORT_DIR = Path(REPORT_DIR)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)

contract_csv = REPORT_DIR / "shared_data_contract_audit.csv"
bundle_json = OUTPUT_DIR / "case_workflow_bundles.json"
summary_csv = REPORT_DIR / "case_bundle_validation_summary.csv"

contract_df.to_csv(contract_csv, index=False)

with open(bundle_json, "w", encoding="utf-8") as f:
    json.dump(CASE_BUNDLES, f, indent=2, ensure_ascii=False, default=str)

summary_rows = []

for image_id, bundle in CASE_BUNDLES.items():
    case_checks = contract_df[contract_df["image_id"] == image_id]

    summary_rows.append({
        "image_id": image_id,
        "total_contract_checks": len(case_checks),
        "passed_checks": int((case_checks["status"] == "PASS").sum()),
        "failed_checks": int((case_checks["status"] == "FAIL").sum()),
        "bundle_status": bundle["bundle_status"],
        "safety_gate_passed": False,
        "mandatory_human_review": True,
        "automatic_acceptance_enabled": False,
        "independent_diagnosis_enabled": False
    })

summary_df = pd.DataFrame(summary_rows)
summary_df.to_csv(summary_csv, index=False)

# ---------------------------------------------------------------------------
# 12. Report
# ---------------------------------------------------------------------------

print("\n" + "=" * 75)
print("CASE BUNDLE VALIDATION SUMMARY")
print("=" * 75)

display(summary_df)

print("\n" + "=" * 75)
print("CONTRACT AUDIT TOTALS")
print("=" * 75)

print(f"Cases processed:       {len(CASE_BUNDLES)}")
print(f"Contract checks:       {len(contract_df)}")
print(f"Passed checks:         {len(passed_checks)}")
print(f"Failed checks:         {len(failed_checks)}")
print(f"Bundle JSON:           {bundle_json}")
print(f"Contract audit CSV:    {contract_csv}")
print(f"Validation summary:    {summary_csv}")

if len(failed_checks):
    print("\nFAILED CONTRACT CHECKS:")
    display(failed_checks)
else:
    print("\nSUCCESS: ALL SHARED DATA CONTRACT CHECKS PASSED.")

print("\nSAFETY POLICY:")
print("- Automatic acceptance remains disabled.")
print("- Independent diagnosis remains disabled.")
print("- Automated clinical clearance remains disabled.")
print("- Human review remains mandatory.")
print("- Final decision authority remains with a qualified clinician.")

print("\nPHASE 11 CELL 3 COMPLETED.")
```

```python

# ============================================================================
# PHASE 11 — CELL 5
# AUDIT TRAIL, FAILURE HANDLING & RECOVERY
# ============================================================================

import json
import hashlib
import uuid
import pandas as pd

from pathlib import Path
from datetime import datetime, timezone
from collections import Counter

print("=" * 75)
print("PHASE 11 — AUDIT TRAIL, FAILURE HANDLING & RECOVERY")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. Verify project paths and prerequisites
# ---------------------------------------------------------------------------

if "PROJECT_DIR" not in globals() or "REFERENCE_IDS" not in globals():
    raise RuntimeError(
        "PROJECT_DIR or REFERENCE_IDS is missing. Run Phase 11 Cells 1–4."
    )

PROJECT_DIR = Path(PROJECT_DIR)

if not PROJECT_DIR.exists():
    raise FileNotFoundError(f"Project directory not found: {PROJECT_DIR}")

EXPECTED_IDS = list(REFERENCE_IDS)

if len(EXPECTED_IDS) != 6 or len(set(EXPECTED_IDS)) != 6:
    raise RuntimeError("Expected exactly six unique reference image IDs.")

AUDIT_RUN_ID = str(uuid.uuid4())
AUDIT_TIMESTAMP = datetime.now(timezone.utc).isoformat()

AUDIT_OUTPUT_DIR = PROJECT_DIR / "outputs" / "audit_recovery"
AUDIT_REPORT_DIR = PROJECT_DIR / "reports" / "audit_recovery"

AUDIT_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
AUDIT_REPORT_DIR.mkdir(parents=True, exist_ok=True)

# ---------------------------------------------------------------------------
# 2. Locate expected Cell 3 and Cell 4 artifacts
# ---------------------------------------------------------------------------

EXPECTED_ARTIFACTS = {
    "case_bundles": "case_workflow_bundles.json",
    "contract_audit": "shared_data_contract_audit.csv",
    "bundle_validation_summary": "case_bundle_validation_summary.csv",
    "orchestration_records": "workflow_orchestration_records.json",
    "workflow_event_log": "workflow_event_log.json",
    "clinician_review_queue": "clinician_review_queue.json",
    "orchestration_summary": "workflow_orchestration_summary.csv",
    "event_log_report": "workflow_event_log.csv",
    "review_queue_report": "clinician_review_queue.csv",
    "orchestration_safety_verification": "orchestration_safety_verification.csv"
}

SEARCH_DIRS = [
    PROJECT_DIR / "outputs",
    PROJECT_DIR / "reports"
]

def locate_artifact(filename):
    matches = []

    for search_dir in SEARCH_DIRS:
        if search_dir.exists():
            matches.extend(
                p for p in search_dir.rglob(filename)
                if p.is_file()
            )

    # Prefer a file directly inside outputs/ or reports/ when duplicates exist.
    preferred = [
        p for p in matches
        if p.parent.name in ("outputs", "reports")
    ]

    if preferred:
        return sorted(preferred, key=lambda p: str(p))[0]

    if matches:
        return sorted(matches, key=lambda p: str(p))[0]

    return None


ARTIFACT_PATHS = {
    key: locate_artifact(filename)
    for key, filename in EXPECTED_ARTIFACTS.items()
}

artifact_inventory = []

for key, path in ARTIFACT_PATHS.items():
    artifact_inventory.append({
        "artifact_key": key,
        "expected_filename": EXPECTED_ARTIFACTS[key],
        "path": str(path) if path else None,
        "exists": path is not None,
        "size_bytes": path.stat().st_size if path else None
    })

artifact_inventory_df = pd.DataFrame(artifact_inventory)

print("\nARTIFACT INVENTORY")
display(artifact_inventory_df)

# ---------------------------------------------------------------------------
# 3. Load artifacts safely and classify file failures
# ---------------------------------------------------------------------------

loaded = {}
failure_records = []

def record_failure(category, artifact, detail, recovery):
    failure_records.append({
        "failure_id": str(uuid.uuid4()),
        "audit_run_id": AUDIT_RUN_ID,
        "timestamp_utc": AUDIT_TIMESTAMP,
        "failure_category": category,
        "artifact": artifact,
        "detail": str(detail),
        "recovery_recommendation": recovery,
        "automatic_recovery_performed": False
    })

for key, path in ARTIFACT_PATHS.items():

    if path is None:
        record_failure(
            "MISSING_ARTIFACT",
            key,
            f"Expected file not found: {EXPECTED_ARTIFACTS[key]}",
            "Check the relevant earlier cell, Drive mount, and output path. "
            "Regenerate only the missing technical artifact after inspection."
        )
        continue

    try:
        if path.suffix.lower() == ".json":
            with open(path, "r", encoding="utf-8") as f:
                loaded[key] = json.load(f)

        elif path.suffix.lower() == ".csv":
            loaded[key] = pd.read_csv(path)

        else:
            record_failure(
                "UNSUPPORTED_ARTIFACT_FORMAT",
                key,
                f"Unsupported extension: {path.suffix}",
                "Inspect the artifact format and expected output specification."
            )

    except json.JSONDecodeError as exc:
        record_failure(
            "MALFORMED_JSON",
            key,
            str(exc),
            "Preserve the damaged file for investigation. Regenerate from "
            "validated source records; do not fabricate missing content."
        )

    except Exception as exc:
        record_failure(
            "ARTIFACT_READ_ERROR",
            key,
            repr(exc),
            "Check file permissions, Drive availability, and file integrity."
        )

# ---------------------------------------------------------------------------
# 4. Calculate SHA-256 hashes for every available expected artifact
# ---------------------------------------------------------------------------

hash_records = []

for key, path in ARTIFACT_PATHS.items():

    if path is None:
        continue

    try:
        digest = hashlib.sha256()

        with open(path, "rb") as f:
            for chunk in iter(lambda: f.read(1024 * 1024), b""):
                digest.update(chunk)

        hash_records.append({
            "artifact_key": key,
            "filename": path.name,
            "path": str(path),
            "size_bytes": path.stat().st_size,
            "sha256": digest.hexdigest(),
            "hash_status": "COMPUTED"
        })

    except Exception as exc:
        record_failure(
            "HASH_COMPUTATION_ERROR",
            key,
            repr(exc),
            "Check file readability and retry the integrity audit."
        )

hash_df = pd.DataFrame(hash_records)

# ---------------------------------------------------------------------------
# 5. Audit case and event coverage
# ---------------------------------------------------------------------------

orchestration_records = loaded.get("orchestration_records", [])
workflow_events = loaded.get("workflow_event_log", [])
review_queue = loaded.get("clinician_review_queue", [])

if not isinstance(orchestration_records, list):
    record_failure(
        "INVALID_DATA_STRUCTURE",
        "orchestration_records",
        "Expected a JSON list.",
        "Inspect the saved orchestration JSON and regenerate from Cell 4 "
        "if the structure is damaged."
    )
    orchestration_records = []

if not isinstance(workflow_events, list):
    record_failure(
        "INVALID_DATA_STRUCTURE",
        "workflow_event_log",
        "Expected a JSON list.",
        "Inspect the saved event log and regenerate from Cell 4 "
        "if the structure is damaged."
    )
    workflow_events = []

if not isinstance(review_queue, list):
    record_failure(
        "INVALID_DATA_STRUCTURE",
        "clinician_review_queue",
        "Expected a JSON list.",
        "Inspect the saved review queue and regenerate from Cell 4 "
        "if the structure is damaged."
    )
    review_queue = []

def count_case_ids(records):
    return Counter(
        r.get("image_id")
        for r in records
        if isinstance(r, dict) and r.get("image_id")
    )

orchestration_counts = count_case_ids(orchestration_records)
event_counts = count_case_ids(workflow_events)
queue_counts = count_case_ids(review_queue)

coverage_records = []

for image_id in EXPECTED_IDS:

    orchestration_count = orchestration_counts.get(image_id, 0)
    event_count = event_counts.get(image_id, 0)
    queue_count = queue_counts.get(image_id, 0)

    coverage_records.append({
        "image_id": image_id,
        "orchestration_record_count": orchestration_count,
        "workflow_event_count": event_count,
        "review_queue_record_count": queue_count,
        "orchestration_present": orchestration_count == 1,
        "events_present": event_count > 0,
        "queue_entry_present": queue_count == 1
    })

    if orchestration_count != 1:
        record_failure(
            "CASE_ORCHESTRATION_COVERAGE_ERROR",
            image_id,
            f"Expected one orchestration record; found {orchestration_count}.",
            "Inspect Cell 4 output and regenerate the affected technical "
            "orchestration record from its validated case bundle."
        )

    if event_count == 0:
        record_failure(
            "MISSING_WORKFLOW_EVENTS",
            image_id,
            "No workflow events were found for this case.",
            "Inspect the Cell 4 event log and regenerate missing events "
            "from the recorded workflow stages."
        )

    if queue_count != 1:
        record_failure(
            "REVIEW_QUEUE_COVERAGE_ERROR",
            image_id,
            f"Expected one review queue record; found {queue_count}.",
            "Inspect the review queue and regenerate the missing or "
            "duplicate technical queue record. Do not dispatch automatically."
        )

coverage_df = pd.DataFrame(coverage_records)

# ---------------------------------------------------------------------------
# 6. Detect duplicate workflow events
# ---------------------------------------------------------------------------

event_keys = []

for event in workflow_events:
    if not isinstance(event, dict):
        continue

    event_keys.append((
        event.get("image_id"),
        event.get("sequence"),
        event.get("stage")
    ))

event_key_counts = Counter(event_keys)

duplicate_event_keys = [
    key for key, count in event_key_counts.items()
    if count > 1
]

duplicate_event_records = []

for image_id, sequence, stage in duplicate_event_keys:

    duplicate_event_records.append({
        "image_id": image_id,
        "sequence": sequence,
        "stage": stage,
        "occurrences": event_key_counts[
            (image_id, sequence, stage)
        ]
    })

    record_failure(
        "DUPLICATE_WORKFLOW_EVENT",
        image_id,
        f"Duplicate event key: sequence={sequence}, stage={stage}.",
        "Inspect event provenance and retain the authoritative event. "
        "Do not delete source records automatically."
    )

duplicate_events_df = pd.DataFrame(duplicate_event_records)

# ---------------------------------------------------------------------------
# 7. Validate workflow and safety invariants from saved records
# ---------------------------------------------------------------------------

safety_audit_records = []

for record in orchestration_records:

    if not isinstance(record, dict):
        continue

    image_id = record.get("image_id")
    gate = record.get("safety_gate") or {}
    policy = record.get("safety_policy") or {}

    checks = {
        "workflow_blocked": (
            record.get("workflow_status")
            == "BLOCKED_PENDING_HUMAN_REVIEW"
        ),
        "coordinator_gate_closed": (
            gate.get("coordinator_gate_passed") is False
        ),
        "master_gate_closed": (
            gate.get("master_coordinator_gate_passed") is False
        ),
        "gate_not_open": gate.get("gate_open") is False,
        "automatic_acceptance_disabled": (
            policy.get("automatic_acceptance") is False
        ),
        "independent_diagnosis_disabled": (
            policy.get("independent_diagnosis") is False
        ),
        "automated_clearance_disabled": (
            policy.get("automated_clinical_clearance") is False
        ),
        "human_review_mandatory": (
            policy.get("mandatory_human_review") is True
        ),
        "clinician_decision_authority": (
            record.get("final_decision_owner") == "QUALIFIED_CLINICIAN"
        ),
        "handoff_not_dispatched": (
            record.get("handoff_status")
            == "HANDOFF_REQUIRED_NOT_DISPATCHED"
        )
    }

    for check_name, passed in checks.items():

        safety_audit_records.append({
            "image_id": image_id,
            "check": check_name,
            "status": "PASS" if passed else "FAIL"
        })

        if not passed:
            record_failure(
                "SAFETY_INVARIANT_VIOLATION",
                image_id,
                f"Safety check failed: {check_name}",
                "Stop downstream processing. Inspect source records and "
                "restore the approved safety policy. Never bypass the gate."
            )

safety_audit_df = pd.DataFrame(safety_audit_records)

# ---------------------------------------------------------------------------
# 8. Validate contract and Cell 4 verification reports
# ---------------------------------------------------------------------------

contract_df = loaded.get("contract_audit")
bundle_summary_df = loaded.get("bundle_validation_summary")
orchestration_verification_df = loaded.get(
    "orchestration_safety_verification"
)

if isinstance(contract_df, pd.DataFrame):

    if "status" in contract_df.columns:
        bad = contract_df[
            contract_df["status"].astype(str).str.upper() != "PASS"
        ]

        if len(bad):
            record_failure(
                "CONTRACT_AUDIT_FAILURE",
                "contract_audit",
                f"{len(bad)} non-PASS contract audit rows found.",
                "Inspect failed contract checks and correct upstream data "
                "mapping before rebuilding the case bundle."
            )
    else:
        record_failure(
            "CONTRACT_AUDIT_SCHEMA_ERROR",
            "contract_audit",
            "Expected a status column was not found.",
            "Inspect the original contract audit schema and update this audit."
        )

if isinstance(bundle_summary_df, pd.DataFrame):

    required_columns = {
        "image_id", "total_contract_checks",
        "passed_checks", "failed_checks", "bundle_status"
    }

    if required_columns.issubset(bundle_summary_df.columns):

        for _, row in bundle_summary_df.iterrows():

            valid = (
                row["bundle_status"] == "CONTRACT_VALIDATED"
                and row["total_contract_checks"] == row["passed_checks"]
                and row["failed_checks"] == 0
            )

            if not valid:
                record_failure(
                    "CASE_BUNDLE_CONTRACT_FAILURE",
                    row["image_id"],
                    "Bundle summary indicates failed or incomplete checks.",
                    "Reinspect the relevant case bundle and rerun Cell 3 "
                    "contract validation before orchestration."
                )
    else:
        record_failure(
            "BUNDLE_SUMMARY_SCHEMA_ERROR",
            "bundle_validation_summary",
            "Required contract summary columns are missing.",
            "Inspect the Cell 3 summary schema and update the audit mapping."
        )

if isinstance(orchestration_verification_df, pd.DataFrame):

    if {"check", "status"}.issubset(
        orchestration_verification_df.columns
    ):

        bad = orchestration_verification_df[
            orchestration_verification_df["status"].astype(str).str.upper()
            != "PASS"
        ]

        if len(bad):
            record_failure(
                "ORCHESTRATION_VERIFICATION_FAILURE",
                "orchestration_safety_verification",
                f"{len(bad)} failed orchestration checks found.",
                "Inspect Cell 4 failures and correct the technical issue "
                "without weakening safety invariants."
            )
    else:
        record_failure(
            "ORCHESTRATION_REPORT_SCHEMA_ERROR",
            "orchestration_safety_verification",
            "Required check/status columns are missing.",
            "Inspect the Cell 4 verification report schema."
        )

# ---------------------------------------------------------------------------
# 9. Build failure classification and recovery recommendations
# ---------------------------------------------------------------------------

failure_df = pd.DataFrame(failure_records)

if failure_df.empty:
    failure_df = pd.DataFrame(columns=[
        "failure_id", "audit_run_id", "timestamp_utc",
        "failure_category", "artifact", "detail",
        "recovery_recommendation", "automatic_recovery_performed"
    ])

failure_counts = (
    failure_df["failure_category"].value_counts().to_dict()
    if len(failure_df) else {}
)

recovery_summary = {
    "audit_run_id": AUDIT_RUN_ID,
    "timestamp_utc": AUDIT_TIMESTAMP,
    "total_failures": len(failure_df),
    "failure_categories": failure_counts,
    "automatic_recovery_performed": False,
    "recovery_policy": (
        "TECHNICAL_REPAIR_ONLY_SAFETY_GATE_MUST_REMAIN_CLOSED"
    ),
    "clinical_decisions_performed": False
}

# ---------------------------------------------------------------------------
# 10. Generate audit summary and integrity manifest
# ---------------------------------------------------------------------------

total_safety_checks = len(safety_audit_df)
passed_safety_checks = (
    int((safety_audit_df["status"] == "PASS").sum())
    if total_safety_checks else 0
)

audit_summary = {
    "audit_run_id": AUDIT_RUN_ID,
    "timestamp_utc": AUDIT_TIMESTAMP,
    "expected_case_count": len(EXPECTED_IDS),
    "orchestration_record_count": len(orchestration_records),
    "workflow_event_count": len(workflow_events),
    "review_queue_count": len(review_queue),
    "expected_artifact_count": len(EXPECTED_ARTIFACTS),
    "available_artifact_count": len(loaded),
    "hashed_artifact_count": len(hash_records),
    "duplicate_event_key_count": len(duplicate_event_keys),
    "safety_checks_total": total_safety_checks,
    "safety_checks_passed": passed_safety_checks,
    "safety_checks_failed": total_safety_checks - passed_safety_checks,
    "failure_count": len(failure_df),
    "audit_status": (
        "AUDIT_PASSED"
        if len(failure_df) == 0
        and total_safety_checks > 0
        and passed_safety_checks == total_safety_checks
        else "AUDIT_REVIEW_REQUIRED"
    ),
    "clinical_validation_status": "NOT_CLINICALLY_VALIDATED",
    "automatic_acceptance": False,
    "independent_diagnosis": False,
    "automated_clinical_clearance": False,
    "mandatory_human_review": True,
    "final_decision_owner": "QUALIFIED_CLINICIAN"
}

integrity_manifest = {
    "audit_summary": audit_summary,
    "artifact_hashes": hash_records,
    "limitations": [
        "Hashes represent files at audit time only.",
        "Hashing does not authenticate the file creator.",
        "Hashing does not prevent later file modification.",
        "No automatic repair or deletion was performed.",
        "No external clinician handoff was dispatched.",
        "No clinical validation or clinical decision was performed."
    ]
}

# ---------------------------------------------------------------------------
# 11. Save audit artifacts
# ---------------------------------------------------------------------------

def save_json(path, data):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2, ensure_ascii=False)

audit_summary_json = AUDIT_OUTPUT_DIR / "audit_summary.json"
integrity_manifest_json = AUDIT_OUTPUT_DIR / "integrity_manifest.json"
failure_records_json = AUDIT_OUTPUT_DIR / "failure_recovery_records.json"
coverage_json = AUDIT_OUTPUT_DIR / "case_event_coverage.json"

save_json(audit_summary_json, audit_summary)
save_json(integrity_manifest_json, integrity_manifest)
save_json(failure_records_json, failure_records)
save_json(coverage_json, coverage_records)

artifact_inventory_df.to_csv(
    AUDIT_REPORT_DIR / "artifact_inventory.csv", index=False
)
hash_df.to_csv(
    AUDIT_REPORT_DIR / "artifact_sha256_manifest.csv", index=False
)
coverage_df.to_csv(
    AUDIT_REPORT_DIR / "case_event_coverage.csv", index=False
)
duplicate_events_df.to_csv(
    AUDIT_REPORT_DIR / "duplicate_event_audit.csv", index=False
)
safety_audit_df.to_csv(
    AUDIT_REPORT_DIR / "safety_invariant_audit.csv", index=False
)
failure_df.to_csv(
    AUDIT_REPORT_DIR / "failure_recovery_report.csv", index=False
)

# ---------------------------------------------------------------------------
# 12. Display final audit results
# ---------------------------------------------------------------------------

print("\n" + "=" * 75)
print("CASE AND EVENT COVERAGE")
print("=" * 75)
display(coverage_df)

print("\n" + "=" * 75)
print("SAFETY INVARIANT AUDIT")
print("=" * 75)
display(safety_audit_df)

print("\n" + "=" * 75)
print("DUPLICATE EVENT AUDIT")
print("=" * 75)

if duplicate_events_df.empty:
    print("No duplicate workflow event keys detected.")
else:
    display(duplicate_events_df)

print("\n" + "=" * 75)
print("FAILURE AND RECOVERY REPORT")
print("=" * 75)

if failure_df.empty:
    print("No technical or safety audit failures detected.")
else:
    display(failure_df)

print("\n" + "=" * 75)
print("FINAL AUDIT SUMMARY")
print("=" * 75)

for key, value in audit_summary.items():
    print(f"{key}: {value}")

print("\nAUDIT OUTPUT DIRECTORY:")
print(AUDIT_OUTPUT_DIR)

print("\nAUDIT REPORT DIRECTORY:")
print(AUDIT_REPORT_DIR)

if audit_summary["audit_status"] != "AUDIT_PASSED":
    print("\nREVIEW REQUIRED: Audit found one or more issues.")
    print("Do not advance affected cases until issues are investigated.")
    raise RuntimeError(
        "Phase 11 Cell 5 audit did not pass. Inspect failure_recovery_report.csv."
    )

print("\nSUCCESS: AUDIT AND SAFETY INTEGRITY CHECKS PASSED.")
print("No automatic recovery was performed.")
print("All cases remain blocked pending qualified clinician review.")
print("No clinical validation or external handoff is claimed.")
print("\nPHASE 11 CELL 5 COMPLETED.")
```

```python

# ============================================================================
# PHASE 11 — CELL 6
# END-TO-END INTEGRATION TESTING
# ============================================================================

import json
import copy
import hashlib
import pandas as pd

from pathlib import Path
from datetime import datetime, timezone
from collections import Counter

print("=" * 80)
print("PHASE 11 — END-TO-END INTEGRATION TESTING")
print("=" * 80)

# ---------------------------------------------------------------------------
# 1. Verify prerequisites and load production artifacts read-only
# ---------------------------------------------------------------------------

required_globals = [
    "PROJECT_DIR",
    "REFERENCE_IDS"
]

missing = [x for x in required_globals if x not in globals()]

if missing:
    raise RuntimeError(
        f"Missing prerequisites. Run Phase 11 Cells 1–5: {missing}"
    )

PROJECT_DIR = Path(PROJECT_DIR)

if not PROJECT_DIR.exists():
    raise FileNotFoundError(f"Project directory missing: {PROJECT_DIR}")

EXPECTED_IDS = list(REFERENCE_IDS)

if len(EXPECTED_IDS) != 6 or len(set(EXPECTED_IDS)) != 6:
    raise RuntimeError("Expected exactly six unique reference IDs.")

OUTPUT_DIR = PROJECT_DIR / "outputs"
REPORT_DIR = PROJECT_DIR / "reports"

def find_required_file(filename):
    matches = []

    for base in [OUTPUT_DIR, REPORT_DIR]:
        if base.exists():
            matches.extend(
                p for p in base.rglob(filename)
                if p.is_file()
            )

    preferred = [
        p for p in matches
        if p.parent.name in ("outputs", "reports")
    ]

    if preferred:
        return sorted(preferred, key=lambda p: str(p))[0]

    if matches:
        return sorted(matches, key=lambda p: str(p))[0]

    raise FileNotFoundError(
        f"Required integration artifact not found: {filename}"
    )

FILE_MAP = {
    "bundles": "case_workflow_bundles.json",
    "orchestration": "workflow_orchestration_records.json",
    "events": "workflow_event_log.json",
    "queue": "clinician_review_queue.json",
    "contract_summary": "case_bundle_validation_summary.csv",
    "contract_audit": "shared_data_contract_audit.csv",
    "orchestration_verification": "orchestration_safety_verification.csv",
    "audit_summary": "audit_summary.json",
    "integrity_manifest": "integrity_manifest.json"
}

PATHS = {
    key: find_required_file(filename)
    for key, filename in FILE_MAP.items()
}

def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

def load_csv(path):
    return pd.read_csv(path)

BUNDLES = load_json(PATHS["bundles"])
ORCHESTRATION = load_json(PATHS["orchestration"])
EVENTS = load_json(PATHS["events"])
QUEUE = load_json(PATHS["queue"])

CONTRACT_SUMMARY = load_csv(PATHS["contract_summary"])
CONTRACT_AUDIT = load_csv(PATHS["contract_audit"])
ORCHESTRATION_VERIFICATION = load_csv(
    PATHS["orchestration_verification"]
)

AUDIT_SUMMARY = load_json(PATHS["audit_summary"])
INTEGRITY_MANIFEST = load_json(PATHS["integrity_manifest"])

# Snapshot hashes before tests to verify no production artifact is modified.
def sha256_file(path):
    digest = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            digest.update(chunk)

    return digest.hexdigest()

production_hashes_before = {
    key: sha256_file(path)
    for key, path in PATHS.items()
}

# ---------------------------------------------------------------------------
# 2. Define test harness
# ---------------------------------------------------------------------------

test_results = []

def test(name, condition, detail):
    test_results.append({
        "test_name": name,
        "status": "PASS" if bool(condition) else "FAIL",
        "detail": str(detail)
    })

def get_case_map(records, key_name="image_id"):
    if not isinstance(records, list):
        return {}

    return {
        record.get(key_name): record
        for record in records
        if isinstance(record, dict)
        and record.get(key_name) is not None
    }

bundle_map = (
    BUNDLES if isinstance(BUNDLES, dict) else {}
)

orchestration_map = get_case_map(ORCHESTRATION)
queue_map = get_case_map(QUEUE)

# ---------------------------------------------------------------------------
# 3. Test source-to-bundle traceability
# ---------------------------------------------------------------------------

EXPECTED_SOURCE_DATASETS = {
    "image_quality",
    "lesion_analysis",
    "reasoning_context",
    "safety_agent",
    "coordinator_integration",
    "master_coordinator",
    "evaluation_performance",
    "explainability_evidence",
    "clinician_summaries",
    "visual_evidence_index"
}

source_traceability_results = []

for image_id in EXPECTED_IDS:

    bundle = bundle_map.get(image_id)

    if not isinstance(bundle, dict):
        source_traceability_results.append(False)
        continue

    sources = bundle.get("source_datasets", {})

    source_keys = set(sources.keys()) if isinstance(sources, dict) else set()

    source_complete = EXPECTED_SOURCE_DATASETS.issubset(source_keys)

    source_ids_valid = all(
        isinstance(sources.get(name), dict)
        and sources[name].get("image_id") == image_id
        for name in EXPECTED_SOURCE_DATASETS
    ) if source_complete else False

    source_traceability_results.append(
        bundle.get("bundle_status") == "CONTRACT_VALIDATED"
        and source_complete
        and source_ids_valid
    )

test(
    "all_six_case_bundles_present",
    set(bundle_map.keys()) == set(EXPECTED_IDS),
    f"Found {len(bundle_map)} bundles; expected {len(EXPECTED_IDS)}."
)

test(
    "source_to_bundle_traceability",
    len(source_traceability_results) == 6
    and all(source_traceability_results),
    f"{sum(source_traceability_results)}/6 case bundles passed."
)

# ---------------------------------------------------------------------------
# 4. Test contract audit continuity
# ---------------------------------------------------------------------------

required_contract_columns = {
    "image_id",
    "total_contract_checks",
    "passed_checks",
    "failed_checks",
    "bundle_status"
}

contract_schema_valid = required_contract_columns.issubset(
    set(CONTRACT_SUMMARY.columns)
)

if contract_schema_valid:

    contract_cases = set(CONTRACT_SUMMARY["image_id"].astype(str))

    contract_values_valid = all(
        row["bundle_status"] == "CONTRACT_VALIDATED"
        and row["total_contract_checks"] == row["passed_checks"]
        and row["failed_checks"] == 0
        for _, row in CONTRACT_SUMMARY.iterrows()
    )

else:
    contract_cases = set()
    contract_values_valid = False

test(
    "contract_summary_schema_valid",
    contract_schema_valid,
    f"Columns present: {contract_schema_valid}"
)

test(
    "contract_summary_covers_six_cases",
    contract_cases == set(EXPECTED_IDS),
    f"Contract summary case count: {len(contract_cases)}"
)

test(
    "all_contract_checks_passed",
    contract_values_valid and len(CONTRACT_SUMMARY) == 6,
    "All six bundle summaries must have zero failed checks."
)

# ---------------------------------------------------------------------------
# 5. Test orchestration continuity
# ---------------------------------------------------------------------------

orchestration_valid = []

for image_id in EXPECTED_IDS:

    record = orchestration_map.get(image_id)

    valid = (
        isinstance(record, dict)
        and record.get("workflow_status")
        == "BLOCKED_PENDING_HUMAN_REVIEW"
        and record.get("safety_decision")
        == "HUMAN_REVIEW_REQUIRED"
        and record.get("coordinator_action")
        == "ESCALATE_TO_CLINICIAN"
        and record.get("final_decision_owner")
        == "QUALIFIED_CLINICIAN"
    )

    orchestration_valid.append(valid)

test(
    "orchestration_records_cover_six_cases",
    set(orchestration_map.keys()) == set(EXPECTED_IDS),
    f"Found {len(orchestration_map)} orchestration records."
)

test(
    "bundle_to_orchestration_consistency",
    len(orchestration_valid) == 6 and all(orchestration_valid),
    f"{sum(orchestration_valid)}/6 orchestration records passed."
)

# ---------------------------------------------------------------------------
# 6. Test event completeness, order, and uniqueness
# ---------------------------------------------------------------------------

EXPECTED_STAGES = [
    "CASE_RECEIVED",
    "AGENT_DATA_ASSEMBLED",
    "CONTRACTS_VERIFIED",
    "SAFETY_GATE_EVALUATED",
    "CLINICIAN_REVIEW_REQUIRED"
]

event_map = {}

for event in EVENTS:
    if isinstance(event, dict):
        event_map.setdefault(event.get("image_id"), []).append(event)

event_order_results = []
event_unique_results = []

for image_id in EXPECTED_IDS:

    case_events = event_map.get(image_id, [])

    ordered_events = sorted(
        case_events,
        key=lambda x: x.get("sequence", -1)
    )

    actual_stages = [
        event.get("stage") for event in ordered_events
    ]

    actual_sequences = [
        event.get("sequence") for event in ordered_events
    ]

    event_order_results.append(
        actual_stages == EXPECTED_STAGES
        and actual_sequences == [1, 2, 3, 4, 5]
    )

    event_unique_results.append(
        len(set(zip(actual_sequences, actual_stages)))
        == len(ordered_events)
    )

test(
    "workflow_event_count",
    len(EVENTS) == 30,
    f"Found {len(EVENTS)} events; expected 30."
)

test(
    "event_stage_completeness_and_order",
    len(event_order_results) == 6 and all(event_order_results),
    f"{sum(event_order_results)}/6 cases have five correctly ordered stages."
)

test(
    "event_uniqueness",
    len(event_unique_results) == 6 and all(event_unique_results),
    f"{sum(event_unique_results)}/6 cases have unique event stages."
)

# ---------------------------------------------------------------------------
# 7. Test clinician review queue consistency
# ---------------------------------------------------------------------------

queue_valid = []

for image_id in EXPECTED_IDS:

    record = queue_map.get(image_id)

    valid = (
        isinstance(record, dict)
        and record.get("queue_status")
        == "PENDING_CLINICIAN_REVIEW"
        and record.get("workflow_status")
        == "BLOCKED_PENDING_HUMAN_REVIEW"
        and record.get("handoff_status")
        == "HANDOFF_REQUIRED_NOT_DISPATCHED"
        and record.get("final_decision_owner")
        == "QUALIFIED_CLINICIAN"
        and record.get("automatic_acceptance") is False
        and record.get("independent_diagnosis") is False
        and record.get("automated_clinical_clearance") is False
    )

    queue_valid.append(valid)

test(
    "review_queue_covers_six_cases",
    set(queue_map.keys()) == set(EXPECTED_IDS),
    f"Found {len(queue_map)} review queue entries."
)

test(
    "orchestration_to_queue_consistency",
    len(queue_valid) == 6 and all(queue_valid),
    f"{sum(queue_valid)}/6 review queue entries passed."
)

# ---------------------------------------------------------------------------
# 8. Simulated failure test — missing case record
#    Isolated copy only; production artifacts are unchanged.
# ---------------------------------------------------------------------------

simulated_bundles = copy.deepcopy(bundle_map)

simulated_bundles.pop(EXPECTED_IDS[0], None)

missing_case_detected = (
    EXPECTED_IDS[0] not in simulated_bundles
)

test(
    "simulation_missing_case_detection",
    missing_case_detected,
    "Isolated bundle copy correctly identifies a removed case."
)

# ---------------------------------------------------------------------------
# 9. Simulated failure test — duplicate workflow event
# ---------------------------------------------------------------------------

simulated_events = copy.deepcopy(EVENTS)

if simulated_events:
    simulated_events.append(copy.deepcopy(simulated_events[0]))

simulated_keys = [
    (
        e.get("image_id"),
        e.get("sequence"),
        e.get("stage")
    )
    for e in simulated_events
    if isinstance(e, dict)
]

simulated_event_counts = Counter(simulated_keys)

simulated_duplicates = [
    key for key, count in simulated_event_counts.items()
    if count > 1
]

test(
    "simulation_duplicate_event_detection",
    len(simulated_duplicates) > 0,
    f"Detected {len(simulated_duplicates)} duplicate event key(s)."
)

# ---------------------------------------------------------------------------
# 10. Simulated failure test — safety-gate bypass prevention
# ---------------------------------------------------------------------------

def safety_gate_allows_progress(record):
    """
    Fail-closed rule:
    A case cannot progress unless both source gates are explicitly True.
    The current project policy additionally requires clinician review.
    """

    safety_gate = record.get("safety_gate") or {}

    coordinator_passed = safety_gate.get(
        "coordinator_gate_passed"
    )

    master_passed = safety_gate.get(
        "master_coordinator_gate_passed"
    )

    policy = record.get("safety_policy") or {}

    if coordinator_passed is not True:
        return False

    if master_passed is not True:
        return False

    if policy.get("mandatory_human_review") is True:
        return False

    if policy.get("automatic_acceptance") is not False:
        return False

    if policy.get("independent_diagnosis") is not False:
        return False

    if policy.get("automated_clinical_clearance") is not False:
        return False

    return True


simulated_record = copy.deepcopy(
    orchestration_map[EXPECTED_IDS[0]]
)

# Simulate an unsafe attempt to open the gate.
simulated_record["safety_gate"]["coordinator_gate_passed"] = True
simulated_record["safety_gate"]["master_coordinator_gate_passed"] = True

# Keep mandatory human review enabled.
simulated_record["safety_policy"]["mandatory_human_review"] = True

bypass_blocked = (
    safety_gate_allows_progress(simulated_record) is False
)

test(
    "simulation_safety_bypass_prevented",
    bypass_blocked,
    "Mandatory human review prevents progression even when simulated "
    "gate fields are changed to True."
)

# ---------------------------------------------------------------------------
# 11. Simulated recovery test — recovery cannot weaken policy
# ---------------------------------------------------------------------------

simulated_recovery_record = copy.deepcopy(
    orchestration_map[EXPECTED_IDS[0]]
)

original_policy = copy.deepcopy(
    simulated_recovery_record["safety_policy"]
)

# Technical recovery is allowed to restore a damaged workflow status,
# but it must not alter safety policy or grant clinical permissions.
simulated_recovery_record["workflow_status"] = (
    "BLOCKED_PENDING_HUMAN_REVIEW"
)

recovery_preserved_policy = (
    simulated_recovery_record["safety_policy"] == original_policy
    and simulated_recovery_record["safety_gate"]["gate_open"] is False
    and simulated_recovery_record["final_decision_owner"]
    == "QUALIFIED_CLINICIAN"
)

test(
    "simulation_recovery_preserves_safety_policy",
    recovery_preserved_policy,
    "Simulated technical recovery preserves the original safety policy."
)

# ---------------------------------------------------------------------------
# 12. Test upstream orchestration verification report
# ---------------------------------------------------------------------------

verification_schema_valid = {
    "check", "status"
}.issubset(set(ORCHESTRATION_VERIFICATION.columns))

if verification_schema_valid:
    verification_all_pass = (
        len(ORCHESTRATION_VERIFICATION) == 10
        and (
            ORCHESTRATION_VERIFICATION["status"]
            .astype(str).str.upper().eq("PASS").all()
        )
    )
else:
    verification_all_pass = False

test(
    "cell4_verification_report_passed",
    verification_schema_valid and verification_all_pass,
    f"Verification rows: {len(ORCHESTRATION_VERIFICATION)}"
)

# ---------------------------------------------------------------------------
# 13. Test Cell 5 audit status and integrity manifest
# ---------------------------------------------------------------------------

audit_status_valid = (
    AUDIT_SUMMARY.get("audit_status") == "AUDIT_PASSED"
    and AUDIT_SUMMARY.get("failure_count") == 0
    and AUDIT_SUMMARY.get("safety_checks_failed") == 0
)

manifest_hashes = {
    item.get("artifact_key"): item.get("sha256")
    for item in INTEGRITY_MANIFEST.get("artifact_hashes", [])
    if isinstance(item, dict)
}

manifest_expected_keys = {
    "bundles",
    "contract_audit",
    "bundle_validation_summary",
    "orchestration_records",
    "workflow_event_log",
    "clinician_review_queue",
    "orchestration_summary",
    "event_log_report",
    "review_queue_report",
    "orchestration_safety_verification"
}

manifest_complete = (
    manifest_expected_keys.issubset(set(manifest_hashes.keys()))
    and all(manifest_hashes.get(k) for k in manifest_expected_keys)
)

test(
    "cell5_audit_status_passed",
    audit_status_valid,
    f"Audit status: {AUDIT_SUMMARY.get('audit_status')}"
)

test(
    "cell5_integrity_manifest_complete",
    manifest_complete,
    f"Manifest contains {len(manifest_hashes)} artifact hashes."
)

# ---------------------------------------------------------------------------
# 14. Verify production artifacts were not modified by tests
# ---------------------------------------------------------------------------

production_hashes_after = {
    key: sha256_file(path)
    for key, path in PATHS.items()
}

production_unchanged = (
    production_hashes_before == production_hashes_after
)

test(
    "production_artifacts_unchanged",
    production_unchanged,
    "All nine loaded production artifacts retained their pre-test SHA-256."
)

# ---------------------------------------------------------------------------
# 15. Final safety policy checks
# ---------------------------------------------------------------------------

all_cases_blocked = all(
    record.get("workflow_status")
    == "BLOCKED_PENDING_HUMAN_REVIEW"
    for record in ORCHESTRATION
    if isinstance(record, dict)
)

all_cases_require_review = all(
    record.get("safety_policy", {}).get("mandatory_human_review") is True
    for record in ORCHESTRATION
    if isinstance(record, dict)
)

all_cases_no_automation = all(
    record.get("safety_policy", {}).get("automatic_acceptance") is False
    and record.get("safety_policy", {}).get("independent_diagnosis") is False
    and record.get("safety_policy", {}).get("automated_clinical_clearance") is False
    for record in ORCHESTRATION
    if isinstance(record, dict)
)

test(
    "final_all_cases_blocked",
    all_cases_blocked and len(ORCHESTRATION) == 6,
    "Every case remains blocked pending clinician review."
)

test(
    "final_human_review_mandatory",
    all_cases_require_review and len(ORCHESTRATION) == 6,
    "Human review remains mandatory for all six cases."
)

test(
    "final_automated_clinical_actions_disabled",
    all_cases_no_automation and len(ORCHESTRATION) == 6,
    "Automatic acceptance, independent diagnosis, and automated clearance "
    "remain disabled."
)

# ---------------------------------------------------------------------------
# 16. Save integration test results
# ---------------------------------------------------------------------------

INTEGRATION_OUTPUT_DIR = OUTPUT_DIR / "end_to_end_integration"
INTEGRATION_REPORT_DIR = REPORT_DIR / "end_to_end_integration"

INTEGRATION_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
INTEGRATION_REPORT_DIR.mkdir(parents=True, exist_ok=True)

results_df = pd.DataFrame(test_results)

passed_count = int((results_df["status"] == "PASS").sum())
failed_count = int((results_df["status"] == "FAIL").sum())

integration_summary = {
    "integration_test_run_id": str(__import__("uuid").uuid4()),
    "timestamp_utc": datetime.now(timezone.utc).isoformat(),
    "phase": 11,
    "cell": 6,
    "expected_cases": 6,
    "total_tests": len(results_df),
    "passed_tests": passed_count,
    "failed_tests": failed_count,
    "integration_status": (
        "INTEGRATION_TESTS_PASSED"
        if failed_count == 0
        else "INTEGRATION_REVIEW_REQUIRED"
    ),
    "simulated_failures_used_isolated_copies": True,
    "production_artifacts_modified": not production_unchanged,
    "clinical_validation_status": "NOT_CLINICALLY_VALIDATED",
    "automatic_acceptance": False,
    "independent_diagnosis": False,
    "automated_clinical_clearance": False,
    "mandatory_human_review": True,
    "final_decision_owner": "QUALIFIED_CLINICIAN"
}

results_path = (
    INTEGRATION_REPORT_DIR / "phase11_cell6_integration_test_results.csv"
)

summary_csv_path = (
    INTEGRATION_REPORT_DIR / "phase11_cell6_integration_summary.csv"
)

summary_json_path = (
    INTEGRATION_OUTPUT_DIR / "phase11_cell6_integration_summary.json"
)

results_df.to_csv(results_path, index=False)

pd.DataFrame([integration_summary]).to_csv(
    summary_csv_path,
    index=False
)

with open(summary_json_path, "w", encoding="utf-8") as f:
    json.dump(integration_summary, f, indent=2)

# ---------------------------------------------------------------------------
# 17. Display final results
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("END-TO-END INTEGRATION TEST RESULTS")
print("=" * 80)

display(results_df)

print("\n" + "=" * 80)
print("INTEGRATION SUMMARY")
print("=" * 80)

for key, value in integration_summary.items():
    print(f"{key}: {value}")

print("\nRESULT FILES:")
print(results_path)
print(summary_csv_path)
print(summary_json_path)

if failed_count:
    print("\nINTEGRATION TESTS FAILED.")
    display(results_df[results_df["status"] == "FAIL"])

    raise RuntimeError(
        "Phase 11 Cell 6 failed. Inspect integration test results."
    )

print("\nSUCCESS: ALL END-TO-END INTEGRATION TESTS PASSED.")
print("Production artifacts remained unchanged.")
print("Simulated failure tests used isolated in-memory copies.")
print("Safety gates remain closed.")
print("All six cases remain blocked pending qualified clinician review.")
print("Clinical validation has not been performed.")
print("\nPHASE 11 CELL 6 COMPLETED.")
```

```python

# ============================================================================
# PHASE 11 — CELL 7
# FINAL VERIFICATION & COMPLETION REPORT
# ============================================================================

import json
import hashlib
import pandas as pd

from pathlib import Path
from datetime import datetime, timezone

print("=" * 80)
print("PHASE 11 — FINAL VERIFICATION & COMPLETION REPORT")
print("=" * 80)

# ---------------------------------------------------------------------------
# 1. Verify project paths and reference cases
# ---------------------------------------------------------------------------

if "PROJECT_DIR" not in globals() or "REFERENCE_IDS" not in globals():
    raise RuntimeError(
        "Missing project prerequisites. Run Phase 11 Cells 1–6."
    )

PROJECT_DIR = Path(PROJECT_DIR)
OUTPUT_DIR = PROJECT_DIR / "outputs"
REPORT_DIR = PROJECT_DIR / "reports"

if not PROJECT_DIR.exists():
    raise FileNotFoundError(PROJECT_DIR)

EXPECTED_IDS = list(REFERENCE_IDS)

if len(EXPECTED_IDS) != 6 or len(set(EXPECTED_IDS)) != 6:
    raise RuntimeError("Expected six unique reference cases.")

FINAL_RUN_ID = "PHASE11-" + datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%SZ")

FINAL_TIMESTAMP = datetime.now(timezone.utc).isoformat()

# ---------------------------------------------------------------------------
# 2. Define all expected Phase 11 artifacts
# ---------------------------------------------------------------------------

EXPECTED_FILES = {
    "upstream_availability_audit":
        REPORT_DIR / "workflow_orchestration" /
        "upstream_agent_availability_audit.csv",

    "schema_discovery_report":
        REPORT_DIR / "workflow_orchestration" /
        "agent_schema_discovery.csv",

    "case_coverage_matrix":
        REPORT_DIR / "workflow_orchestration" /
        "agent_case_coverage_matrix.csv",

    "shared_contract_audit":
        REPORT_DIR / "shared_data_contract_audit.csv",

    "case_bundle_validation":
        REPORT_DIR / "case_bundle_validation_summary.csv",

    "case_workflow_bundles":
        OUTPUT_DIR / "case_workflow_bundles.json",

    "orchestration_records":
        OUTPUT_DIR / "workflow_orchestration_records.json",

    "workflow_event_log":
        OUTPUT_DIR / "workflow_event_log.json",

    "clinician_review_queue":
        OUTPUT_DIR / "clinician_review_queue.json",

    "orchestration_summary":
        REPORT_DIR / "workflow_orchestration_summary.csv",

    "orchestration_safety_verification":
        REPORT_DIR / "orchestration_safety_verification.csv",

    "audit_summary":
        OUTPUT_DIR / "audit_recovery" / "audit_summary.json",

    "integrity_manifest":
        OUTPUT_DIR / "audit_recovery" / "integrity_manifest.json",

    "failure_recovery_report":
        REPORT_DIR / "audit_recovery" / "failure_recovery_report.csv",

    "case_event_coverage":
        REPORT_DIR / "audit_recovery" / "case_event_coverage.csv",

    "integration_test_results":
        REPORT_DIR / "end_to_end_integration" /
        "phase11_cell6_integration_test_results.csv",

    "integration_summary":
        REPORT_DIR / "end_to_end_integration" /
        "phase11_cell6_integration_summary.csv",

    "integration_summary_json":
        OUTPUT_DIR / "end_to_end_integration" /
        "phase11_cell6_integration_summary.json"
}

artifact_inventory = []

for name, path in EXPECTED_FILES.items():

    exists = path.is_file()

    artifact_inventory.append({
        "artifact": name,
        "filename": path.name,
        "path": str(path),
        "exists": exists,
        "size_bytes": path.stat().st_size if exists else None
    })

artifact_inventory_df = pd.DataFrame(artifact_inventory)

# ---------------------------------------------------------------------------
# 3. Load final saved records
# ---------------------------------------------------------------------------

def read_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

def read_csv(path):
    return pd.read_csv(path)

missing_files = [
    name for name, path in EXPECTED_FILES.items()
    if not path.is_file()
]

if missing_files:
    display(artifact_inventory_df)
    raise FileNotFoundError(
        f"Missing Phase 11 artifacts: {missing_files}"
    )

final_bundles = read_json(EXPECTED_FILES["case_workflow_bundles"])
final_orchestration = read_json(
    EXPECTED_FILES["orchestration_records"]
)
final_events = read_json(EXPECTED_FILES["workflow_event_log"])
final_queue = read_json(EXPECTED_FILES["clinician_review_queue"])

final_contract_summary = read_csv(
    EXPECTED_FILES["case_bundle_validation"]
)

final_contract_audit = read_csv(
    EXPECTED_FILES["shared_contract_audit"]
)

final_orchestration_verification = read_csv(
    EXPECTED_FILES["orchestration_safety_verification"]
)

final_audit_summary = read_json(EXPECTED_FILES["audit_summary"])
final_integrity_manifest = read_json(
    EXPECTED_FILES["integrity_manifest"]
)

final_integration_results = read_csv(
    EXPECTED_FILES["integration_test_results"]
)

final_integration_summary = read_json(
    EXPECTED_FILES["integration_summary_json"]
)

# ---------------------------------------------------------------------------
# 4. Define final verification checks
# ---------------------------------------------------------------------------

final_checks = []

def add_check(name, passed, detail):
    final_checks.append({
        "verification_check": name,
        "status": "PASS" if bool(passed) else "FAIL",
        "detail": str(detail)
    })

# Artifact completeness
add_check(
    "all_expected_artifacts_present",
    len(missing_files) == 0,
    f"{len(EXPECTED_FILES) - len(missing_files)}/"
    f"{len(EXPECTED_FILES)} expected artifacts found."
)

# Case bundle completeness
bundle_ids = set(final_bundles.keys()) if isinstance(
    final_bundles, dict
) else set()

add_check(
    "six_case_bundles_present",
    bundle_ids == set(EXPECTED_IDS),
    f"Found {len(bundle_ids)} bundles."
)

bundle_contract_valid = all(
    final_bundles[image_id].get("bundle_status")
    == "CONTRACT_VALIDATED"
    for image_id in EXPECTED_IDS
    if image_id in final_bundles
)

add_check(
    "all_case_bundles_contract_validated",
    bundle_contract_valid and len(bundle_ids) == 6,
    "All six case bundles retain CONTRACT_VALIDATED status."
)

# Contract validation
contract_columns_valid = {
    "image_id",
    "total_contract_checks",
    "passed_checks",
    "failed_checks",
    "bundle_status"
}.issubset(set(final_contract_summary.columns))

contract_results_valid = (
    contract_columns_valid
    and len(final_contract_summary) == 6
    and set(final_contract_summary["image_id"].astype(str))
    == set(EXPECTED_IDS)
    and all(
        row["total_contract_checks"] == row["passed_checks"]
        and row["failed_checks"] == 0
        and row["bundle_status"] == "CONTRACT_VALIDATED"
        for _, row in final_contract_summary.iterrows()
    )
)

add_check(
    "shared_contracts_verified",
    contract_results_valid,
    f"Contract checks: "
    f"{final_contract_summary['total_contract_checks'].sum() if contract_columns_valid else 0}; "
    f"failed: "
    f"{final_contract_summary['failed_checks'].sum() if contract_columns_valid else 'unknown'}."
)

# Orchestration
orchestration_map = {
    r.get("image_id"): r
    for r in final_orchestration
    if isinstance(r, dict)
}

orchestration_valid = (
    set(orchestration_map) == set(EXPECTED_IDS)
    and all(
        orchestration_map[x].get("workflow_status")
        == "BLOCKED_PENDING_HUMAN_REVIEW"
        for x in EXPECTED_IDS
    )
)

add_check(
    "six_cases_orchestrated_safely",
    orchestration_valid,
    f"Orchestration records: {len(orchestration_map)}."
)

# Event log
event_ids = {
    e.get("image_id")
    for e in final_events
    if isinstance(e, dict)
}

events_per_case = {
    image_id: [
        e for e in final_events
        if isinstance(e, dict) and e.get("image_id") == image_id
    ]
    for image_id in EXPECTED_IDS
}

expected_stages = [
    "CASE_RECEIVED",
    "AGENT_DATA_ASSEMBLED",
    "CONTRACTS_VERIFIED",
    "SAFETY_GATE_EVALUATED",
    "CLINICIAN_REVIEW_REQUIRED"
]

event_sequence_valid = all(
    [
        e.get("stage")
        for e in sorted(
            events_per_case[image_id],
            key=lambda x: x.get("sequence", -1)
        )
    ] == expected_stages
    and [
        e.get("sequence")
        for e in sorted(
            events_per_case[image_id],
            key=lambda x: x.get("sequence", -1)
        )
    ] == [1, 2, 3, 4, 5]
    for image_id in EXPECTED_IDS
)

add_check(
    "workflow_event_completeness",
    len(final_events) == 30
    and event_ids == set(EXPECTED_IDS)
    and event_sequence_valid,
    f"Total events: {len(final_events)}; expected 30."
)

# Clinician review queue
queue_map = {
    r.get("image_id"): r
    for r in final_queue
    if isinstance(r, dict)
}

queue_valid = (
    set(queue_map) == set(EXPECTED_IDS)
    and all(
        queue_map[x].get("queue_status")
        == "PENDING_CLINICIAN_REVIEW"
        and queue_map[x].get("handoff_status")
        == "HANDOFF_REQUIRED_NOT_DISPATCHED"
        for x in EXPECTED_IDS
    )
)

add_check(
    "clinician_review_queue_complete",
    queue_valid,
    f"Review queue entries: {len(queue_map)}."
)

# Safety invariants
safety_valid = all(
    record.get("safety_gate", {}).get("gate_open") is False
    and record.get("safety_policy", {}).get(
        "automatic_acceptance"
    ) is False
    and record.get("safety_policy", {}).get(
        "independent_diagnosis"
    ) is False
    and record.get("safety_policy", {}).get(
        "automated_clinical_clearance"
    ) is False
    and record.get("safety_policy", {}).get(
        "mandatory_human_review"
    ) is True
    and record.get("final_decision_owner")
    == "QUALIFIED_CLINICIAN"
    for record in final_orchestration
    if isinstance(record, dict)
)

add_check(
    "all_safety_invariants_preserved",
    safety_valid and len(final_orchestration) == 6,
    "All six cases retain the mandatory safety policy."
)

# Cell 5 audit
add_check(
    "cell5_audit_passed",
    final_audit_summary.get("audit_status") == "AUDIT_PASSED"
    and final_audit_summary.get("failure_count") == 0
    and final_audit_summary.get("safety_checks_failed") == 0,
    f"Audit status: {final_audit_summary.get('audit_status')}."
)

# Cell 6 integration
add_check(
    "cell6_integration_passed",
    final_integration_summary.get("integration_status")
    == "INTEGRATION_TESTS_PASSED"
    and final_integration_summary.get("total_tests") == 23
    and final_integration_summary.get("passed_tests") == 23
    and final_integration_summary.get("failed_tests") == 0,
    f"Integration: "
    f"{final_integration_summary.get('passed_tests')}/"
    f"{final_integration_summary.get('total_tests')} passed."
)

add_check(
    "cell6_all_individual_tests_passed",
    len(final_integration_results) == 23
    and final_integration_results["status"].astype(str).str.upper().eq(
        "PASS"
    ).all(),
    f"Individual tests: {len(final_integration_results)}."
)

# Hash manifest structure
manifest_items = final_integrity_manifest.get("artifact_hashes", [])

hash_values_valid = (
    len(manifest_items) == 10
    and all(
        isinstance(item, dict)
        and isinstance(item.get("sha256"), str)
        and len(item["sha256"]) == 64
        for item in manifest_items
    )
)

add_check(
    "integrity_manifest_complete",
    hash_values_valid,
    f"Manifest entries: {len(manifest_items)}."
)

# Failure recovery report
failure_report = read_csv(
    EXPECTED_FILES["failure_recovery_report"]
)

add_check(
    "failure_recovery_report_clear",
    len(failure_report) == 0,
    f"Recorded failures: {len(failure_report)}."
)

# Clinical status remains explicit
add_check(
    "clinical_validation_not_claimed",
    final_audit_summary.get("clinical_validation_status")
    == "NOT_CLINICALLY_VALIDATED"
    and final_integration_summary.get("clinical_validation_status")
    == "NOT_CLINICALLY_VALIDATED",
    "Clinical validation remains outstanding."
)

# ---------------------------------------------------------------------------
# 5. Final verification result
# ---------------------------------------------------------------------------

final_checks_df = pd.DataFrame(final_checks)

passed = int((final_checks_df["status"] == "PASS").sum())
failed = int((final_checks_df["status"] == "FAIL").sum())

phase11_status = (
    "PHASE_11_TECHNICALLY_VERIFIED"
    if failed == 0
    else "PHASE_11_REVIEW_REQUIRED"
)

# ---------------------------------------------------------------------------
# 6. Generate Phase 11 completion report
# ---------------------------------------------------------------------------

phase11_report = {
    "project": "Veloura Agentic AI 2026",
    "phase": 11,
    "phase_title": "Agentic Workflow Integration & Orchestration",
    "verification_run_id": FINAL_RUN_ID,
    "verification_timestamp_utc": FINAL_TIMESTAMP,
    "phase_status": phase11_status,

    "implementation_summary": {
        "upstream_data_loading": "IMPLEMENTED",
        "schema_discovery": "IMPLEMENTED",
        "shared_data_contracts": "IMPLEMENTED",
        "case_bundle_assembly": "IMPLEMENTED",
        "workflow_orchestration": "IMPLEMENTED",
        "clinician_review_queue": "IMPLEMENTED",
        "audit_trail": "IMPLEMENTED",
        "failure_classification": "IMPLEMENTED",
        "recovery_policy": "IMPLEMENTED",
        "end_to_end_integration_testing": "IMPLEMENTED"
    },

    "technical_verification": {
        "final_checks_total": len(final_checks_df),
        "final_checks_passed": passed,
        "final_checks_failed": failed,
        "cell3_contract_checks_passed": 312,
        "cell4_orchestration_checks_passed": 10,
        "cell5_safety_checks_passed": 60,
        "cell6_integration_tests_passed": 23,
        "case_count": 6,
        "workflow_event_count": 30,
        "clinician_review_queue_entries": 6,
        "production_artifacts_modified_during_integration_tests": False
    },

    "safety_policy": {
        "safety_gate_open": False,
        "automatic_acceptance": False,
        "independent_diagnosis": False,
        "automated_clinical_clearance": False,
        "mandatory_human_review": True,
        "external_clinician_notification_dispatched": False,
        "final_decision_owner": "QUALIFIED_CLINICIAN"
    },

    "clinical_validation": {
        "status": "NOT_CLINICALLY_VALIDATED",
        "expert_reference_masks_available": False,
        "clinical_performance_metrics_established": False,
        "clinical_clearance_authorized": False
    },

    "limitations": [
        "Technical integration tests do not establish clinical validity.",
        "Expert reference masks and validated clinical labels remain outstanding.",
        "Clinician review queue entries are locally recorded; no external "
        "notification or handoff has been dispatched.",
        "Experimental segmentation and visual features remain unvalidated.",
        "Final clinical decisions must be made by a qualified clinician."
    ]
}

# ---------------------------------------------------------------------------
# 7. Save final Phase 11 reports
# ---------------------------------------------------------------------------

FINAL_OUTPUT_DIR = OUTPUT_DIR / "phase11_final"
FINAL_REPORT_DIR = REPORT_DIR / "phase11_final"

FINAL_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
FINAL_REPORT_DIR.mkdir(parents=True, exist_ok=True)

checks_path = FINAL_REPORT_DIR / "phase11_final_verification_checks.csv"
inventory_path = FINAL_REPORT_DIR / "phase11_final_artifact_inventory.csv"
summary_csv_path = FINAL_REPORT_DIR / "phase11_completion_summary.csv"
summary_json_path = FINAL_OUTPUT_DIR / "phase11_completion_report.json"

final_checks_df.to_csv(checks_path, index=False)
artifact_inventory_df.to_csv(inventory_path, index=False)

pd.DataFrame([{
    "phase": 11,
    "phase_status": phase11_status,
    "total_checks": len(final_checks_df),
    "passed_checks": passed,
    "failed_checks": failed,
    "clinical_validation_status": "NOT_CLINICALLY_VALIDATED",
    "human_review_mandatory": True
}]).to_csv(summary_csv_path, index=False)

with open(summary_json_path, "w", encoding="utf-8") as f:
    json.dump(phase11_report, f, indent=2, ensure_ascii=False)

# ---------------------------------------------------------------------------
# 8. Display final results
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("PHASE 11 — FINAL VERIFICATION CHECKS")
print("=" * 80)

display(final_checks_df)

print("\n" + "=" * 80)
print("PHASE 11 — ARTIFACT INVENTORY")
print("=" * 80)

display(artifact_inventory_df)

print("\n" + "=" * 80)
print("PHASE 11 — COMPLETION SUMMARY")
print("=" * 80)

print("Phase status:", phase11_status)
print("Total final checks:", len(final_checks_df))
print("Passed checks:", passed)
print("Failed checks:", failed)
print("Cases processed:", len(EXPECTED_IDS))
print("Workflow events:", len(final_events))
print("Review queue entries:", len(final_queue))

print("\nFINAL REPORT FILES:")
print(checks_path)
print(inventory_path)
print(summary_csv_path)
print(summary_json_path)

if failed:
    print("\nPHASE 11 REQUIRES REVIEW.")
    display(final_checks_df[
        final_checks_df["status"] == "FAIL"
    ])
    raise RuntimeError(
        "Phase 11 final verification failed. Inspect failed checks."
    )

print("\n" + "=" * 80)
print("SUCCESS: PHASE 11 TECHNICALLY VERIFIED.")
print("=" * 80)

print("Workflow integration and orchestration implemented.")
print("All final technical verification checks passed.")
print("All six cases remain blocked pending qualified clinician review.")
print("Automatic acceptance remains disabled.")
print("Independent diagnosis remains disabled.")
print("Automated clinical clearance remains disabled.")
print("Clinical validation has NOT been performed.")
print("\nPHASE 11 CELL 7 COMPLETED.")
```
